### Inspecting the Uploaded Files
Let's print the contents of the example input and output files to understand their structure.

````python
def preview_file(filepath, lines_to_show=10):
    print(f"=== {filepath} (First {lines_to_show} lines) ===")
    try:
        with open(filepath, 'r') as f:
            for i in range(lines_to_show):
                line = f.readline()
                if not line:
                    break
                print(line.rstrip('\n'))
    except Exception as e:
        print(f"Error reading file: {e}")
    print("=" * 40 + "\n")

# Preview the example input and output files
preview_file('in_level-1_0-example.txt')
preview_file('out_level-1_0-example.txt')
````

### Extracting Instructions from `level-1.pdf`
Let's read the text content of the PDF to understand the algorithm we need to implement.

````python
!pip install -q pypdf

import pypdf

def extract_pdf_text(filepath):
    print(f"=== Extracting Text from {filepath} ===")
    try:
        reader = pypdf.PdfReader(filepath)
        for i, page in enumerate(reader.pages):
            print(f"--- Page {i + 1} ---")
            print(page.extract_text())
    except Exception as e:
        print(f"Error reading PDF: {e}")

extract_pdf_text('level-1.pdf')
````

### Solving Level 1
We will write a python script to parse the input files, apply the formula $\text{Desks} = (x // 3) * y$, and save the outputs to text files.

````python
import glob

def solve_level_1(input_file):
    output_file = input_file.replace('in_', 'out_')
    print(f"Processing: {input_file} -> {output_file}")

    with open(input_file, 'r') as infile:
        lines = infile.read().strip().split('\n')

    num_rooms = int(lines[0])
    results = []

    for i in range(1, num_rooms + 1):
        x, y = map(int, lines[i].split())
        # Max desks is (x / 3) * y
        desks = (x // 3) * y
        results.append(str(desks))

    with open(output_file, 'w') as outfile:
        outfile.write('\n'.join(results) + '\n')
    print(f"Successfully wrote {len(results)} outputs to {output_file}")

# Solve for all matching input files in the folder
input_files = sorted(glob.glob('in_level-1_*.txt'))
for file in input_files:
    solve_level_1(file)
````
